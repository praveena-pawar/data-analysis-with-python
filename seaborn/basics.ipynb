```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
print(type(tips))
tips.head()
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker",
    size="size",
    style="smoker"
)
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="line",
    hue="smoker",
    size="size",
    style="smoker"
)
```

```python
# scatter plot

tips = sns.load_dataset("tips")
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time"
)
```

```python
# Line plot

tips = sns.load_dataset("tips")
sns.lineplot(
    data=tips,
    x="total_bill",
    y="tip",
)
```

```python
flights = sns.load_dataset("flights")

sns.lineplot(
    data=flights,
    x="year",
    y="passengers"
)
```

```python
flights.head()
```

```python
# Bar plots
print(tips.head())

sns.barplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)
```

```python
# Box plot

sns.boxplot(
    data=tips,
    x="day",
    y="tip"
)
```

```python
# Histogram 

penguins = sns.load_dataset("penguins")

sns.histplot(
    data=penguins,
    x="body_mass_g",
    bins=20
)
```

```python
penguins.head()
```

```python
# Heat maps

flights
```

```python
flights_pivot = flights.pivot(index="month", columns="year", values="passengers")

sns.heatmap(
    data=flights_pivot,
    cmap="coolwarm"
)
```

```python
flights_pivot = flights.pivot(index="month", columns="year", values="passengers")

sns.heatmap(
    data=flights_pivot,
    cmap="coolwarm",
    annot=True,
    fmt="d"
)

plt.title("Passengers Heatmap")
plt.tight_layout()
```

```python
fig, ax = plt.subplots()

sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    marker="o",
    ax=ax
)

ax.set_title("avg bill values for each day")
ax.set_xlabel("Days")
ax.set_ylabel("Total Bill Amount")

fig.tight_layout()
```
