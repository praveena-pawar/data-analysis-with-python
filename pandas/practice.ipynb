```python
import pandas as pd

data = {
    "order_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "customer": ["A", "B", "A", "C", "B", "A", "C", "B"],
    "product": ["Laptop", "Mobile", "Tablet", "Laptop", "Tablet", "Mobile", "Mobile", "Laptop"],
    "price": [50000, 20000, 15000, 52000, 14000, 21000, 19000, 51000],
    "quantity": [1, 2, 1, 1, 3, 1, 2, 1],
    "date": pd.to_datetime([
        "2024-01-01", "2024-01-02", "2024-01-03", "2024-01-04",
        "2024-01-05", "2024-01-06", "2024-01-07", "2024-01-08"
    ])
}

df = pd.DataFrame(data)
```
