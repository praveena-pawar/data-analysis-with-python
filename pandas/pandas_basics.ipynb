```python
import pandas as pd
```

```python
s = pd.Series([1, 2, 3, 4])
print(s)
print(type(s))
```

```python
#Indexing 
print(s[0])
print(s[2])
```

```python
# Label 
s = pd.Series([20, 25, 22, 24], index=["Adam", "Bob", "Eve", "Charlie"])
print(s)
```

```python
#DataFrame
info = {
    "Name" : ["Adam", "Bob", "Eve"],
    "Age" : [20, 22, 24],
    "Marks" : [85, 92, 75]
}

df = pd.DataFrame(info)
df
```

```python
print(df.index)
print(df.columns)
```
