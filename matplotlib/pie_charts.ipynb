```python
import matplotlib.pyplot as plt
```

```python
# Pie Charts 

expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

plt.pie(amounts, labels=expenses)
plt.style.available
plt.style.use('default')
```

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

plt.pie(amounts, labels=expenses, autopct="%1.1f%%") #for values 



expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

plt.pie(amounts, labels=expenses, autopct="%1.1f%%",
        wedgeprops={
            "edgecolor":"black",
            "linewidth":2,
            "linestyle":"--"
        }
       
       )
```

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

plt.pie(amounts, labels=expenses, autopct="%1.1f%%",
        wedgeprops={
            "edgecolor":"black",
            "linewidth":1,
            "linestyle":"--",
            "fill":False
        }
       
       )
```

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

explod = [0, 0, 0, 0.2, 0]

plt.pie(amounts, labels=expenses, autopct="%1.1f%%",
        wedgeprops={
            "edgecolor":"black",
            "linewidth":1,
            "linestyle":"--"
        },
        explode=explod,
        startangle=360
       
       )
```
