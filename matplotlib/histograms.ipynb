```python
import matplotlib.pyplot as plt
```

```python
import numpy as np

np.random.seed(0)

scores = np.random.normal(70, 10, 100)

plt.hist(scores)
```

```python
np.random.seed(0)

scores = np.random.normal(70, 10, 100)

plt.hist(scores, color="skyblue", edgecolor="black")
plt.xlabel("Scores")
plt.ylabel("#No of students")
```

```python
np.random.seed(0)

scores = np.random.normal(70, 10, 100)

bins = [0, 30, 70, 90]

plt.hist(scores, color="skyblue", edgecolor="black", bins=bins)

plt.xlabel("Scores")
plt.ylabel("#No of students")
plt.title("Distribution of marks of students")
print(scores)
```

```python
np.random.seed(0)

scores = np.random.normal(70, 10, 100)

bins = [0, 30, 70, 90]

plt.hist(scores, color="skyblue", edgecolor="black", bins=bins)

plt.axvline(
    x=33,
    label="Passing marks(33)",
    color="red",
    linewidth="2",
    linestyle="--"
)

plt.xlabel("Scores")
plt.ylabel("#No of students")
plt.title("Distribution of marks of students")
print(scores)
```
