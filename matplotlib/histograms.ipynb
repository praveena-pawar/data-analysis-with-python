```python
import matplotlib.pyplot as plt
```

```python
import numpy as np

np.random.seed(0)

scores = np.random.normal(70, 10, 100)

plt.hist(scores)
```

```python
np.random.seed(0)

scores = np.random.normal(70, 10, 100)

plt.hist(scores, color="skyblue", edgecolor="black")
plt.xlabel("Scores")
plt.ylabel("#No of students")
```
