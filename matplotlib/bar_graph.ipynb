```python
import numpy as np
oscar_movies = [
    "Ther Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "the Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M

non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
# non_oscar_years = [2008, 2009, 2010, 2011, 2012,]
non_oscar_revenue = [378, 2788, 829, 185, 275] # in $M

x = np.arange(len(years))
width = 0.4

plt.bar(x-width/2, oscar_revenue, width, label="Osacr Movies") #Shifting to LEFT
plt.bar(x+width/2, non_oscar_revenue, width, label="Non-Osacr Movies") # Shifting to RIGHT

plt.title("Oscar VS Non oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.xticks(x, years)
```
