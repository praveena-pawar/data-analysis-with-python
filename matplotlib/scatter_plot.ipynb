```python
import matplotlib.pyplot as plt
```

```python
# Scatter_Plots 

people = ["Person A", "Person B", "Person C","Person D", "Person E", 
          "Person F", "Person G", "Person H", "Person I", "Person J",]

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pre = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

colors = ["green" if x < 135 else "red" for x in blood_pre]

plt.scatter(age, blood_pre, color=colors, alpha=0.9)
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood pressure")
plt.grid()
plt.colorbar(label="BP")

for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i], blood_pre[i]), xytext=(age[i]+1, blood_pre[i]))

plt.xlim(min(age), max(age)+10)
plt.ylim(min(blood_pre), max(blood_pre)+5)
```
